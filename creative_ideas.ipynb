## 2. Enter your API key securely

When prompted, paste your OpenAI API key. `getpass` hides what you type. Do not put your key in a notebook cell, save it in the notebook, or share it.

```python
from getpass import getpass
from openai import OpenAI

# The key is entered at runtime and is not written into this notebook.
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)
```

## 3. Choose an everyday object

Type an object you see or use in daily life, such as a mug, paperclip, or umbrella.

```python
everyday_object = input("Enter an everyday object: ").strip()

if not everyday_object:
    raise ValueError("Please enter an object before continuing.")
```

## 4. Ask for one creative idea

The next cell sends your object to the OpenAI Responses API and prints one unusual, practical idea.

```python
response = client.responses.create(
    model="gpt-6-luna",
    input=(
        f"Suggest ONE unusual, creative, and practical use for this everyday object: "
        f"{everyday_object}. Give a brief explanation."
    ),
)

print(response.output_text)
```
